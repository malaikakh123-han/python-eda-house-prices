```python
Q1_price = df['price_numeric'].quantile(0.25)
Q3_price = df['price_numeric'].quantile(0.75)

Q1_price, Q3_price
```

```python
IQR_price = Q3_price - Q1_price

IQR_price
```

```python
lower_bound_price = Q1_price - 1.5 * IQR_price
upper_bound_price = Q3_price + 1.5 * IQR_price

print("Lower Bound:", lower_bound_price)
print("Upper Bound:", upper_bound_price)
```

```python
price_outliers = df[
    (df['price_numeric'] < lower_bound_price) |
    (df['price_numeric'] > upper_bound_price)
]

price_outliers.shape
```

```python
price_outliers[
    ['property_id', 'city', 'property_type', 'area_sqft',
     'bedrooms', 'bathrooms', 'price_numeric']
].sort_values('price_numeric', ascending=False)
```

```python
Q1_area = df['area_sqft'].quantile(0.25)
Q3_area = df['area_sqft'].quantile(0.75)

IQR_area = Q3_area - Q1_area

print("Q1:", Q1_area)
print("Q3:", Q3_area)
print("IQR:", IQR_area)
```

```python
lower_bound_area = Q1_area - 1.5 * IQR_area
upper_bound_area = Q3_area + 1.5 * IQR_area

print("Lower Bound:", lower_bound_area)
print("Upper Bound:", upper_bound_area)
```

```python
area_outliers = df[
    (df['area_sqft'] < lower_bound_area) |
    (df['area_sqft'] > upper_bound_area)
]

area_outliers.shape
```

```python
area_outliers[
    ['property_id', 'city', 'property_type', 'area_sqft',
     'bedrooms', 'bathrooms', 'price_numeric']
].sort_values('area_sqft', ascending=False)
```

```python
price_mean = df['price_numeric'].mean()
price_std = df['price_numeric'].std()

df['price_zscore'] = (
    (df['price_numeric'] - price_mean) / price_std
)

df[['price_numeric', 'price_zscore']].head()
```

```python
zscore_outliers = df[
    df['price_zscore'].abs() > 3
]

zscore_outliers.shape
```

```python
zscore_outliers[
    ['property_id', 'city', 'property_type',
     'area_sqft', 'bedrooms', 'bathrooms',
     'price_numeric', 'price_zscore']
].sort_values('price_numeric', ascending=False)
```




The histogram shows that property prices are concentrated toward the lower end of the distribution, with a long right tail extending toward very high prices. The box plot also shows numerous observations above the main distribution, indicating the presence of potential high-price outliers.

Using the IQR method, the first quartile (Q1) of property price is Rs. 5.75 million and the third quartile (Q3) is Rs. 11.45 million, giving an IQR of Rs. 5.7 million. The resulting upper bound is Rs. 20 million, and 70 observations are identified as potential price outliers. The lower bound is negative, so the IQR method does not classify the zero-price observations as outliers using this rule.

The potential price outliers include properties priced at Rs. 100 million, Rs. 150 million, Rs. 200 million, and Rs. 300 million. Some of these properties have relatively small areas compared with their prices, so they require further investigation. They may represent genuine high-value properties or possible data-entry or pricing issues. They should not be automatically removed.

For property area, Q1 is 1,290.5 square feet and Q3 is 2,165.5 square feet, producing an IQR of 875 square feet. The upper bound is 3,478 square feet, and 17 observations are identified as potential area outliers. Several observations have an area of 50,000 square feet, which is substantially above the upper IQR boundary and should be investigated further.

The Z-score method identified 15 property-price observations with an absolute Z-score greater than 3. The most extreme observations have Z-scores of approximately 20.47. The IQR method identified more potential price outliers than the Z-score method.

The IQR method is useful because it is based on quartiles and is less affected by extreme values. However, it can identify a relatively large number of observations as outliers in a highly skewed dataset. The Z-score method measures how far an observation is from the mean in terms of standard deviations, but it can be strongly influenced by extreme values and is less suitable when the distribution is highly skewed. Therefore, both methods should be used as indicators for investigation rather than automatic rules for deleting observations.

Task 7 Interpretation
1. Correlation Analysis

The correlation matrix shows that the numerical variables have generally weak linear relationships with property price. Among the explanatory variables, bathrooms have the highest positive correlation with price at approximately 0.15, followed by bedrooms at approximately 0.07 and area_sqft at approximately 0.05.

Parking spaces, year built, and floors have correlations with price close to zero, while distance from the center has a very weak negative correlation of approximately -0.02.

Overall, none of the numerical variables shows a strong linear correlation with property price in this dataset. This suggests that property price is not well explained by any single numerical variable through a simple linear relationship. The presence of extreme values and data-quality issues identified earlier may also affect these correlations.

2. Correlation Heatmap

The heatmap confirms the generally weak relationships between property price and the other numerical variables. The strongest relationship among the explanatory variables is between bedrooms and bathrooms, with a correlation of approximately 0.39. Area and bathrooms also show a positive correlation of approximately 0.24.

The correlations between most other numerical variables are close to zero. Therefore, there is no evidence of strong multicollinearity among most of the numerical variables.

 3. Multivariate Analysis

The multivariate scatter plot examines property area and price while using city as an additional variable. The main concentration of observations suggests that larger properties generally tend to have higher prices, although the relationship is not strong or perfectly consistent.

The plot also shows several extreme observations with exceptionally high prices and some unusually large property areas. The different city groups overlap substantially, indicating that area and city together do not produce a simple separation of property prices.

These results indicate that property price is likely influenced by multiple factors rather than one numerical variable alone. However, the analysis shows association rather than causation, so these relationships should not be interpreted as evidence that one variable directly causes changes in price.

8.1 Key Findings

Finding 1 — Property prices are highly right-skewed

The distribution of property prices is strongly right-skewed. Most properties are concentrated in the lower price range, while a relatively small number of properties have extremely high prices. The mean price is approximately PKR 9.96 million, while the median is approximately PKR 8.40 million, indicating that high-priced observations pull the mean upward.

Finding 2 — The dataset contains substantial categorical inconsistencies

Several categorical variables contain inconsistent representations. For example, the city variable contains different capitalization formats such as Karachi, KARACHI, and karachi. The furnished variable contains multiple representations including Yes, yes, YES, Y, True, No, no, NO, N, and False. These inconsistencies can cause the same category to be counted separately and can affect analysis.

Finding 3 — Karachi has the largest number of listings

After standardizing the capitalization of city names, Karachi has the largest number of listings, followed by Lahore and Islamabad. The remaining cities have smaller numbers of observations. This indicates that the dataset is not evenly distributed across cities.

Finding 4 — Houses are the most common property type

After standardizing capitalization, houses represent the largest property-type category, followed by apartments, villas, and townhouses. Residential houses and flats occur much less frequently in comparison.

Finding 5 — Numerical variables have weak correlations with price

The correlation analysis shows that none of the numerical variables has a strong linear correlation with property price. Bathrooms have the highest correlation at approximately 0.15, while area has a correlation of approximately 0.05. The weak correlations suggest that property price cannot be adequately explained by one numerical variable alone.

8.2 Recommendations

Recommendation 1 — Standardize categorical variables

Before performing further analysis or building predictive models, categorical variables should be standardized. City names should use consistent capitalization, while furnishing values such as Yes, YES, Y, and True should be mapped to a common representation. This will prevent duplicate categories and improve the reliability of future analysis.

Recommendation 2 — Investigate and appropriately handle extreme values

The analysis identified substantial outliers in property price and area. These observations should be investigated individually to determine whether they represent genuine luxury properties, unusual listings, or data-entry errors. They should not automatically be removed without understanding their source.

Recommendation 3 — Use multiple variables for future price analysis

Because the numerical variables individually show weak correlations with price, future analysis should consider multiple factors simultaneously. Additional variables such as city, property type, furnishing status, condition, and other relevant property characteristics could be incorporated into a more comprehensive analysis or predictive model.
